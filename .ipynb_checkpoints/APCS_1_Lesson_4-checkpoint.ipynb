## APCS 8-2 最短距離

題目說明：
狼人會躲在某位村民家裡，占卜師會告訴你狼人的位置，請判斷村民的
位置是否會被狼人攻擊

A~B 為村民全家，下方數值則為住家與住家之間的距離

```
A(loc=0) B    C      D      E          F(loc=5)     G
|--------|----|------|------|----------|------------|
    4      2     3      3        5           6
```

輸入說明：

- 一共有 2 行，第一行有 3 個數字
- 第 1 行 3 個數字分表代表「村民位置」、「狼人位置」、「狼人的攻擊範圍」
- 第 2 行數字代表每個住家與左邊住家的距離，第 1 棟就是 0

輸出說明：

- `Yes` 找表會被攻擊，`No` 代表不會。

範例輸入：

```
1 5 10
0 4 2 3 3 5 6
```

應得輸出：

```
No
```


````python
from typing import Literal

# import sys
# s = sys.stdin.read().split()

def parser(s: str) -> (str, str, str, [str]):
    villager, wolfman, attack_range, _, *house_dist = map(int, s.split())
    return villager, wolfman, attack_range, house_dist

def is_attack(s: str) -> Literal["Yes", "No"]:
    villager, wolfman, attack_range, house_dist = parser(s)
    begin, end = (villager, wolfman) if wolfman > villager else (wolfman, villager)
    vw_dist = sum(house_dist[begin:end])
    return "Yes" if vw_dist < attack_range else "No"
````

````python
s = """
6 2 5
0 4 2 3 3 5 6
"""
assert is_attack(s) == "No"
s = """
1 5 10
0 4 2 3 3 5 6
"""
assert is_attack(s) == "No"
s = """
3 5 15
0 4 2 3 3 5 6
"""
assert is_attack(s) == "Yes"

print("Test complete")
````
