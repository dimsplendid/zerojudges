## APCS 8-2 最短距離

題目說明：
狼人會躲在某位村民家裡，占卜師會告訴你狼人的位置，請判斷村民的
位置是否會被狼人攻擊

A~B 為村民全家，下方數值則為住家與住家之間的距離

```
A(loc=0) B    C      D      E          F(loc=5)     G
|--------|----|------|------|----------|------------|
    4      2     3      3        5           6
```

輸入說明：

- 一共有 2 行，第一行有 3 個數字
- 第 1 行 3 個數字分表代表「村民位置」、「狼人位置」、「狼人的攻擊範圍」
- 第 2 行數字代表每個住家與左邊住家的距離，第 1 棟就是 0

輸出說明：

- `Yes` 找表會被攻擊，`No` 代表不會。

範例輸入：

```
1 5 10
0 4 2 3 3 5 6
```

應得輸出：

```
No
```


````python
# import sys
# s = sys.stdin.read() # 非 IPO

from typing import Literal

def parse(s: str) -> tuple[int, int, int, list[int]]: # 非 IPO
    villager, wolfman, attack_range, _, *house_dist = map(int, s.split())
    return villager, wolfman, attack_range, house_dist

def is_attack(
    villager: int, 
    wolfman: int, 
    attack_range: int, 
    house_dist: list[int]
) -> Literal["Yes", "No"]:
    
    begin, end = (villager, wolfman) if wolfman > villager else (wolfman, villager)
    vw_dist = sum(house_dist[begin:end])
    
    return "Yes" if vw_dist < attack_range else "No"

def main(args):
    return is_attack(*parse(args))
````

````python
#start

s = """
6 2 5
0 4 2 3 3 5 6
"""
assert main(s) == "No"
s = """
1 5 10
0 4 2 3 3 5 6
"""
assert main(s) == "No"
s = """
3 5 15
0 4 2 3 3 5 6
"""
assert main(s) == "Yes"

print("Test Complete.")
````

## APCS 8-3 自訂函式

題目說明：

使用自訂函式語法 def，建立一個數值相加與文字相加的函式，並進行正確的計算

輸入說明：

- 每一行有三個資料
- 第一項代表數值加法或字串加法
- 後兩項為相加的資料


````python
# import sys
# s = sys.stdin.read() # IPO
from typing import Iterable, TypeVar
# IPO


def parse(s: str) -> Iterable[tuple[int, int] | tuple[str, str]]:
    """
    parse input to list of (str, str) or (int, int)
    """
    def check_type(s: str) -> tuple[int, int] | tuple[str, str]:
        opt, a, b = s.split()
        return (int(a), int(b)) if opt == "1" else (a,b)
        
    return map(check_type, s.splitlines())

def add(t: tuple[int, int] | tuple[str, str]) -> str:
    return str(t[0] + t[1]) # type: ignore # cause mypy sucks

def main(args):
    result = map(add, parse(args))
    return "\n".join(result) + "\n" 
    print("\n".join(result))
````

````python
#start

s = """1 3 5
2 a b
1 10 12
2 P A
2 5 G
"""

assert main(s) == """8
ab
22
PA
5G
"""

print("Test Complete.")
````

## APCS 8-4 溫度轉換

題目說明：

- 溫度有兩種單位，分別是攝氏和華氏，但這兩種溫度當中存在一個轉換的公式，可以讓兩種溫度找到一個對應關係。
- 輸入為攝氏溫度，輸出為華氏溫度。

轉換公式：
$$
F = \frac{9}{5} \times C + 32
$$

````python
def parse(s:str) -> Iterable[float]:
    return map(float, s.split())

def celcius_to_farenheit(c:float) -> str:
    return f"{9/5 * c + 32:.0f}"

# s = sys.stdin.read()
def main(args):
    result = map(celcius_to_farenheit, parse(args))
    return "\n".join(result) + "\n"
    print("\n".join(result))
````

````python
s = """0
15
-5
30
"""

assert main(s) == """32
59
23
86
"""

print("Test Complete.")
````

## APCS 8-5 四則運算
題目說明：
- 計算機是生活中的好幫手，接下來的程式需要寫一個簡單型計算機，輸入兩個數字和加減乘除符號後輸出對應的結果。

計算機的規則如下：
- 由於是簡易計算機，只要進行兩個數字的運算
- 此程式不需要判斷括號
- 四則運算符號如: +, -, *, / 


````python
from enum import Enum

class Operator(Enum):
    ADD = '+'
    SUB = '-'
    MUL = '*'
    DIV = '/'

def parse(xs: str) -> Iterable[tuple[Operator, int, int]]:

    def check_operator(x: str) -> tuple[Operator, int, int]:
        a, op, b = x.split()
        return Operator(op), int(a), int(b)
    
    return map(check_operator, xs.splitlines())

def calculate(x: tuple[Operator, int, int]) -> str:
    op, a, b = x
    match op:
        case Operator.ADD: result = a + b
        case Operator.SUB: result = a - b
        case Operator.MUL: result = a * b
        case Operator.DIV: result = a / b if b != 0 else "N/A" # type: ignore
        case _: raise ValueError("Unreachable")
    return f"{result}"

def main(args):
    result = map(calculate, parse(args))
    return "\n".join(result) + "\n"
    print("\n".join(result))
    
````

````python
s = """4 + 5
2 - 8
12 / 3
7 * 9
13 / 0
"""

assert main(s) == """9
-6
4.0
63
N/A
"""

print("Test Complete.")
````

## APCS 8-6 外部函式

題目說明：
- 輸入圓的半徑，計算圓周長和面積
- 需使用數學模組引入圓周率
- 計算結果請使用 round() 取到小數下第二位
- 此題不需要使用連續 IPO


````python
import math

def parse(s: str) -> float:
    return float(s)
    
def circ_and_area(r: float):
    return 2 * math.pi * r, math.pi * r * r
    
def main(args):
    circ, area = circ_and_area(parse(args))
    return f"{circ:.2f} {area:.2f}" # str(round(...)) 好醜……
````

````python
assert main("5") == "31.42 78.54"
assert main("10") == "62.83 314.16"
assert main("15") == "94.25 706.86"

print("Test Complete.")
````

## APCS 8-7 `random`

題目說明：
設計一個點數為 1~6 點的骰子，輸入為骰子的數量。
輸出有兩行，第一行為各骰子點數，第二行為所有骰子的總和。
- 此題需要使用隨機數模組
- 輸入骰子的數量 $1 \le N \le 10$
> 你的執行輸出「可能」與範例輸出不同


````python
from random import randint

def parse(s: str) -> Iterable[int]:
    return map(int, s.splitlines())

def d6(n:int) -> list[int]:
    return [randint(1, 6) for _ in range(n)]

def main(args):
    results = map(d6, parse(args))
    for result in results:
        print(" ".join(map(str, result)))
        print(sum(result))
````

````python
s = """1
3
5
"""
main(s)
````

## APCS 9-1 `print()` 格式

題目說明：
- 在生活中，有許多文字內容是使用格式化輸出的方式呈現，使用者只需要事先填入對應的數字或字詞，就能產生出一段對話。
    - 例如：公車上的到站提示、鐵路、高鐵的目前位置、大量寄送的電子賀卡等。
- 請你利用這個特性，建立一個字庫和格式化文字，輸出對應的氣象預報文字。

輸入說明：

- 每筆輸入只有一行，分為兩個參數
- 第 1 個為數值，代表目前氣溫
- 第 2 個為文字，代表天氣概況

輸出說明：

- 利用建好的格式，輸出預報結果


````python
def parse(s: str) -> Iterable[tuple[str, str]]:
    return map(str.split, s.splitlines()) # type: ignore

def format_weather(temperature: int, weather: str) -> str:
    return f"今天的氣溫是 {temperature} 度，天氣是{weather}天"

def main(args):
    results = map(lambda s: format_weather(*s), parse(args))
    return "\n".join(results) + "\n"
    print("\n".join(results))
````

## APCS 9-3 split() 與重組

題目說明：
阿妮在一間出版社負責編輯的工作，為了編修文章，常常需要增加或減少標點符號，有天電腦不小心中毒，導致文章的檔案被加入了不需要的符號，請你設計一個程式，將這些符號移除。
- 需要保留的符號： . , !
- 不需要的符號：() - [] {} ; : <>?@#$%^&*

輸入說明：

- 每一行有一筆文字測資

輸出說明：

- 輸出排除掉不需要符號的結果
- 需要保留空格、及指定的符號


````python
def parse(s: str) -> Iterable[str]:
    return s.splitlines()

def remove_symbols(s: str) -> str:
    def check_symbol(c):
        if c in '()-[]{};:<>?@#$%^&*':
            return False
        return True
    return ''.join(filter(check_symbol, s))

def main(args):
    results = map(remove_symbols, parse(args))
    return "\n".join(results) + "\n"
````

````python
s = """he---ll[o], wo?r#l^d!
A(P)-;@C%S&
(Pra#(c)t[ice]-- ?mak(@e&s) perf--ect.
"""

assert main(s) == """hello, world!
APCS
Practice makes perfect.
"""

print("Test Complete.")
````

## APCS 9-4 ASCII

題目說明：

- 英文名字其實隱藏著幸運值，其規則如下
    1.  將英文名字中的所有字元透過 ASCII 轉換為數值
    2. (奇數字元的數值加總 - 偶數字元的數字加總) 取絕對值
    3. 最後再對 5 取餘數

請撰寫一個程式，測試每個英文名的幸運值和代表意思

| 幸運值 | 代表意思  |
|:-----:|:----------|
|   0   |Lucky Ball |
|   1   |Lucky Sheep|
|   2   |Lucky Guy  |
|   3   |OK         |
|   4   |Good Luck  |

輸入說明：

- 每一行為一個英文名字

輸出說明：

- 幸運值所代表的意思


````python
def parser(s:str) -> Iterable[str]:
    return s.splitlines()

lucky_list = ['Lucky Ball', 'Lucky Sheep', 'Lucky Guy', 'OK', 'Good Luck']

def lucky_mean(name:str) -> str:
    ascii_list = [ord(c) for c in name]
    lucky_number = abs(sum([n * (-1) ** i for i, n in enumerate(ascii_list)])) % 5
    return lucky_list[lucky_number]

def main(args):
    result = map(lucky_mean, parser(args))
    return "\n".join(result) + "\n"
````

````python
s = """Kevin
Tony
Leo
Vivian
Charlie
"""

assert main(s) == """Lucky Guy
OK
Lucky Sheep
Good Luck
Lucky Ball
"""

print("Test Complete.")
````

## APCS 9-5 解碼器

題目說明：
- 在密碼學中有一種很簡單的加密方式，就是將明文的每個字元加上某個整數 K 而得到密碼的字元。

輸入說明：

- 每一行為一筆資料，每筆資料分為兩個部分。
- 欲轉換的文字，可能是加密過的結果或一般字詞。
- 第二個為 K 值，用來轉換文字結果， -15 <= K <= 15。

輸出說明：

- 加解密後的字串結果



````python
def parse(s: str) -> Iterable[tuple[str, int]]:
    def unpack(s: str) -> tuple[str, int]:
        txt, shift = s.split()
        return txt, int(shift)
    return map(unpack, s.splitlines())

def encode(s: str, shift: int) -> str:
    result = map(chr, map(lambda n: n+shift, map(ord, s)))
    return "".join(result)

def main(args):
    result = map(lambda t: encode(*t), parse(args))
    return "\n".join(result) + "\n"
````

````python
s = r"""HELLO 3
FUHX -5
Foj^ed 10
Tomorrow 6
`TV[\aX 13
"""

assert main(s) == """KHOOR
APCS
Python
Zusuxxu}
machine
"""

print("Test Complete.")
````
